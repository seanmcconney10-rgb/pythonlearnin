```python

# Create empty list to store expenses
expenses = []

# Each expense will have: category, vendor, amount, date


# --------------------------------------------
# Catcher functions
# --------------------------------------------

# function empty_input(prompt)
#   loop forever
#       get input
#       if input == X -> return None
#       if not empty -> return value
#       else print error

# function non_float(prompt)
#   loop forever
#       try convert to float
#       if works -> return value
#       else print error


# --------------------------------------------
# Adds expenses
# --------------------------------------------

# function add_expense()
#   ask for category, vendor, amount, date
#   if user enters X -> go back
#   store in dictionary
#   append to list


# --------------------------------------------
# View expenses
# --------------------------------------------

# function view_expenses()
#   if list empty -> print message
#   loop through expenses
#   print each expense nicely


# --------------------------------------------
# totals
# --------------------------------------------

# function total_of_all()
#   loop through expenses and sum amounts

# function total_by_category()
#   ask for category (or X to go back)
#   loop and match category

# function total_by_vendor()
#   ask for vendor (or X to go back)
#   loop and match vendor


# --------------------------------------------
# totals menu
# --------------------------------------------

# loop forever
#   show:
#       A all
#       B category
#       C vendor
#       X back
#   run correct function
#   X returns to main menu


# --------------------------------------------
# main loop
# --------------------------------------------

# loop forever
#   show menu:
#       1 add expenses
#       2 view expenses
#       3 totals
#       4 exit
#   run selected option
```

```python

expenses = []


# --------------------------------------------
# catchers
# --------------------------------------------
def empty_input(prompt):
    while True:
        value = input(prompt).strip()

        if value.upper() == "X":
            return None

        if value != "":
            return value

        print("Input is empty: try again! (or X to go back)")


def non_float(prompt):
    while True:
        try:
            return float(input(prompt))

        except:
            print("invalid float: try again!")


# --------------------------------------------
# add expense
# --------------------------------------------
def add_expense():
    print("\n--- Add Expenses ---")

    category = empty_input("Category (or X): ")
    if category is None:
        return

    vendor = empty_input("Vendor (or X): ")
    if vendor is None:
        return

    amount = non_float("Amount: ")

    date = empty_input("Date(year-month-day) (or X): ")
    if date is None:
        return

    expense = {
        "category": category,
        "vendor": vendor,
        "amount": amount,
        "date": date
    }

    expenses.append(expense)
    print("expense added")


# --------------------------------------------
# view expenses
# --------------------------------------------
def view_expenses():
    print("\n--- All Expenses ---")

    if len(expenses) == 0:
        print("No expenses found.")
        return

    for i, es in enumerate(expenses, start=1):
        print(f"\nExpense #{i}")
        print(f"Category: {es['category']}")
        print(f"Vendor:   {es['vendor']}")
        print(f"Amount:   ${es['amount']:.2f}")
        print(f"Date:     {es['date']}")


# --------------------------------------------
# totals
# --------------------------------------------
def total_of_all():
    print("\n--- Total of all Expenses ---")

    total = 0
    for es in expenses:
        total += es["amount"]

    print(f"Total = ${total:.2f}")


def total_by_category():
    print("\n--- Total Expenses by Category ---")

    print()
    category = empty_input("Category (or X): ")
    if category is None:
        return

    total = 0
    for es in expenses:
        if es["category"].lower() == category.lower():
            total += es["amount"]

    print(f"{category} total = ${total:.2f}")


def total_by_vendor():
    print("\n--- Total Expenses by Vendor ---")

    print()
    vendor = empty_input("Vendor (or X): ")
    if vendor is None:
        return

    total = 0
    for es in expenses:
        if es["vendor"].lower() == vendor.lower():
            total += es["amount"]

    print(f"{vendor} total = ${total:.2f}")


# --------------------------------------------
# totals menu
# --------------------------------------------
def totals_menu():
    while True:
        print("\n--- Totals Menu ---")
        print("A. All")
        print("B. Category")
        print("C. Vendor")
        print("X. Back")

        choice = input("Choice(A-C or X): ").upper()

        if choice == "A":
            total_of_all()
        elif choice == "B":
            total_by_category()
        elif choice == "C":
            total_by_vendor()
        elif choice == "X":
            break
        else:
            print("invalid: try again!")


# --------------------------------------------
# main loop
# --------------------------------------------
while True:
    print("\n--- Expenses Tracker ---")
    print("1. Add Expenses")
    print("2. View Expenses")
    print("3. Totals")
    print("4. Exit")

    choice = input("Select(1-4): ")

    if choice == "1":
        add_expense()
    elif choice == "2":
        view_expenses()
    elif choice == "3":
        totals_menu()
    elif choice == "4":
        print("ENDED RUN")
        break
    else:
        print("invalid choice")
```
